```python
import msgpack
import zlib
import pickle

def load_obj(name: str) -> object:
    """
    加载指定名称的 pickle 文件并返回其内容。

    此函数会打开一个以 `.pickle` 为扩展名的文件，并使用 `pickle` 模块加载其内容。

    Args:
        name (str): 不带扩展名的文件名称。

    Returns:
        object: 从 pickle 文件中加载的 Python 对象。

    Raises:
        FileNotFoundError: 如果指定的文件不存在。
        pickle.UnpicklingError: 如果文件内容无法被正确反序列化。
    """
    with open(name + '.pickle', 'rb') as f:
        return pickle.load(f)
def decode_from_bin(file_path):
    """
    从bin文件读取并解码数据，与encode_to_bin函数对应
    
    参数:
        file_path: 要读取的bin文件路径
        
    返回:
        解码后的数据，如果出错则返回None
    """
    try:
        # 1. 读取二进制文件
        with open(file_path, 'rb') as f:
            compressed_data = f.read()
        
        # 2. 解压缩数据（对应编码时的压缩）
        decompressed_data = zlib.decompress(compressed_data)
        
        # 3. 使用msgpack解码数据
        data = msgpack.unpackb(decompressed_data, raw=False)
        
        print(f"数据已从 {file_path} 成功解码")
        return data
        
    except Exception as e:
        print(f"解码过程出错: {str(e)}")
        return None
```
